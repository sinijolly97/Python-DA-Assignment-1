```python
import numpy as np
temperature_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperature_w1)
```

```python
print("shape:", temperature_w1.shape)
print("data type:", temperature_w1.dtype)
print("number of elements:", temperature_w1.size)
```

```python
temperature_w1_f=(temperature_w1*9/5 + 32)
print(temperature_w1_f)
```

```python
min_temp=np.min(temperature_w1)
max_temp=np.max(temperature_w1)
mean_temp=np.mean(temperature_w1)
print(min_temp)
print(max_temp)
print(mean_temp)
```

```python
print("first three days:", temperature_w1[:3])
print("last two days:", temperature_w1[-2:])
print("middle three days:", temperature_w1[2:5])
```

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['rank1', 'rank2', 'rank3', 'rank4', 'rank5']
)
print(marks)
```

```python
print("1st rank student mark:", marks.iloc[0])
```

```python
print("Top 3 ranks marks:", marks.loc['rank1': 'rank3'])
```

```python
print("3d rank student marks:", marks.iloc[2])
```

```python
print("ranks with marks greaterthan 90:", marks[marks>90])
```

```python
marks['rank1']=100
print(marks)
```

```python
marks=marks.drop('rank5')
print(marks)
```

```python
cgpa=marks/10
print(cgpa)
```
